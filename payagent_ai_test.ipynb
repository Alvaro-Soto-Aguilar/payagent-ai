```python
import requests
from google.colab import userdata

# 1. Recuperar las credenciales de forma segura desde los Secretos de Colab
CLIENT_ID = userdata.get('PAYPAL_CLIENT_ID')
SECRET = userdata.get('PAYPAL_SECRET')

print("¡Credenciales cargadas correctamente en Google Colab!")

# 2. Configurar la URL de autenticación de PayPal para el entorno Sandbox
url = "https://api-m.sandbox.paypal.com/v1/oauth2/token"

# 3. Realizar la petición para obtener el token de acceso (OAuth 2.0)
headers = {
    "Accept": "application/json",
    "Accept-Language": "en_US"
}

data = {
    "grant_type": "client_credentials"
}

response = requests.post(url, auth=(CLIENT_ID, SECRET), headers=headers, data=data)

# 4. Comprobar si la conexión con PayPal ha tenido éxito
if response.status_code == 200:
    token_info = response.json()
    print("\n ¡CONEXIÓN EXITOSA CON LA API DE PAYPAL (SANDBOX)!")
    print("Token de acceso obtenido correctamente:")
    print(token_info.get("access_token")[:30] + "...") # Muestra un trozo del token por seguridad
else:
    print(f"\n Error en la conexión: {response.status_code}")
    print(response.text)
```
